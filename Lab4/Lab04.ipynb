### 1. Рядки та Множини
**Постановка задачі (Варіант 3):**
Задано слово. Скласти програму, яка визначає і виводить на екран кількість різних символів в ньому.

**Детальний опис алгоритму та роботи структур даних:**
1. **Ініціалізація:** Отримуємо рядок (слово) за допомогою функції введення та обчислюємо його загальну довжину.
2. **Фільтрація унікальності:** Використовуємо вбудовану структуру даних `set` (множина). Під час перетворення рядка на множину алгоритм автоматично залишає лише унікальні символи, миттєво відкидаючи всі дублікати.
3. **Обчислення результату:** За допомогою функції `len()` визначаємо кількість елементів у сформованій множині, що відповідає кількості різних (унікальних) символів у заданому слові.

```python
word = input("Введіть слово: ")

total_chars = len(word)
unique_chars = set(word)

count_unique = len(unique_chars)

print(f"Загальна кількість символів у слові '{word}': {total_chars}")
print(f"Кількість різних символів: {count_unique}")
print(f"Унікальні символи: {unique_chars}")
```

### 2. Множини (Sets)
**Постановка задачі:**
На кафедрі є три гуртки: програмування, наука про дані та англійська мова. Необхідно згенерувати випадкові списки студентів (по 10 осіб, ID від 1 до 30), перетворити їх на множини та виконати операції над ними (перетин, різниця, об'єднання).

**Детальний опис алгоритму:**
1. **Генерація унікальних значень:** Метод `random.sample` гарантує вибірку унікальних ідентифікаторів без повторень. Згенеровані списки одразу перетворюються на тип `set`.
2. **Операції над множинами:**
   - **Перетин (`&`):** Визначає спільні елементи (студентів, які відвідують декілька гуртків одночасно).
   - **Різниця (`-`):** Відсіює елементи, які присутні в інших множинах, залишаючи лише студентів одного конкретного гуртка.
   - **Об'єднання (`|`):** Зливає всі множини в одну загальну, автоматично видаляючи дублікати, що дозволяє знайти сумарну кількість унікальних студентів.

```python
import random

programming = set(random.sample(range(1, 31), 10))
data_science = set(random.sample(range(1, 31), 10))
english = set(random.sample(range(1, 31), 10))

print(f"Програмування:   {programming}")
print(f"Наука про дані:  {data_science}")
print(f"Англійська мова: {english}\n")

print("1) Відвідують обидва гуртки:")
print(f"- Програмування і наука про дані: {programming & data_science}")
print(f"- Програмування і англійська мова: {programming & english}")
print(f"- Англійська мова і наука про дані: {english & data_science}\n")

all_three = programming & data_science & english
if all_three:
    print(f"2) Відвідують усі три гуртки: {all_three}\n")
else:
    print("2) Відвідують усі три гуртки: Немає таких студентів\n")

only_prog = programming - data_science - english
only_ds = data_science - programming - english
only_eng = english - programming - data_science

only_one = only_prog | only_ds | only_eng
print(f"3) Відвідують лише один гурток: {only_one}\n")

at_least_one = programming | data_science | english
print(f"4) Загальна кількість унікальних студентів: {len(at_least_one)}")
```

### 3. Словники (Dictionaries)
**Постановка задачі:**
Згенерувати 25 унікальних ідентифікаторів студентів. Для кожного згенерувати випадкові оцінки (від 0 до 100) з трьох предметів у вигляді вкладеного словника. Обчислити максимальні та середні бали, перетворити їх у систему ECTS та національну шкалу, вивести таблицю.

**Детальний опис алгоритму:**
1. **Структурування бази даних:** Застосовано модель вкладеного словника. Ключем верхнього рівня є ідентифікатор студента, а значенням — внутрішній словник з предметами та оцінками (складність доступу $O(1)$).
2. **Аналітика через генератори:** Пошук максимального бала та обчислення середніх значень виконується за допомогою генераторних виразів, що є оптимізованим рішенням з точки зору використання пам'яті.
3. **Сортування та форматування:** Словник перетворюється на плоский список кортежів, який сортується за середнім балом у порядку спадання за допомогою `lambda`-функції. Таблиця форматується за допомогою f-рядків з модифікаторами вирівнювання простору.

```python
import random

student_ids = random.sample(range(1, 100001), 25)
subjects = ['Програмування на Python', 'Теорія ймовірностей', 'Іноземна мова']
students_data = {}

for sid in student_ids:
    students_data[sid] = {sub: random.randint(0, 100) for sub in subjects}

selected_subject = 'Програмування на Python'
max_score = max(students_data[sid][selected_subject] for sid in students_data)
top_students = [sid for sid, scores in students_data.items() if scores[selected_subject] == max_score]

print(f"Максимальний бал з '{selected_subject}': {max_score}")
print(f"ID студентів з цим балом: {top_students}\n")

print("Середній бал з кожного предмета:")
for sub in subjects:
    total_score = sum(students_data[sid][sub] for sid in students_data)
    avg_score = total_score / len(students_data)
    print(f"- {sub}: {avg_score:.2f}")

target_student = student_ids[1]
student_scores = students_data[target_student].values()
student_avg = sum(student_scores) / len(student_scores)
print(f"\nСередній бал студента {target_student}: {student_avg:.2f}\n")

def get_grades(score):
    if score >= 90:
        return 'A', 'відмінно'
    elif score >= 82:
        return 'B', 'добре'
    elif score >= 75:
        return 'C', 'добре'
    elif score >= 67:
        return 'D', 'задовільно'
    elif score >= 60:
        return 'E', 'задовільно'
    elif score >= 35:
        return 'FX', 'незадовільно (з можливістю перездачі)'
    else:
        return 'F', "незадовільно"

students_list = []
for sid, scores in students_data.items():
    avg = sum(scores.values()) / len(scores)
    students_list.append((sid, avg, scores))

students_list.sort(key=lambda x: x[1], reverse=True)

print(f"{'ID':<8} | {'Сер.бал':<7} | {'Предмет':<25} | {'Бал':<4} | {'ECTS':<4} | {'Нац. шкала'}")
print("-" * 90)

for sid, avg, scores in students_list:
    first_row = True
    for sub, score in scores.items():
        ects, nat = get_grades(score)
        if first_row:
            print(f"{sid:<8} | {avg:<7.2f} | {sub:<25} | {score:<4} | {ects:<4} | {nat}")
            first_row = False
        else:
            print(f"{'':<8} | {'':<7} | {sub:<25} | {score:<4} | {ects:<4} | {nat}")
    print("-" * 90)
```

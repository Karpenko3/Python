### 1. Робота з одновимірними масивами
**По завданню треба було** створити масив зі 100 випадкових цілих чисел у діапазоні від 0 до 10. Після цього необхідно було знайти всі унікальні числа в цьому масиві, відсортувати їх та порахувати, скільки разів кожне з них зустрічалося.

**Для реалізації ми використали** функцію генерації випадкових чисел `randint` із бібліотеки NumPy для створення початкового набору даних. Щоб не писати складні цикли для пошуку повторень та сортування, ми застосували вбудовану функцію `np.unique`. Вона за один прохід знаходить усі унікальні значення, автоматично їх сортує від найменшого до найбільшого, а завдяки параметру `return_counts=True` ще й повертає масив із частотою їхньої появи.

```python
import numpy as np

arr = np.random.randint(0, 11, 100)
unique_vals, counts = np.unique(arr, return_counts=True)

print("Згенерований масив (100 елементів):")
print(arr)
print("-" * 50)
print("Відсортований масив унікальних значень:")
print(unique_vals)
print("-" * 50)
print("Частота появи кожного значення відповідно:")
print(counts)
```

### Завдання 2. Матриці та базові статистичні операції
**По завданню треба було** згенерувати матрицю розміром 20 на 5 із дійсними числами в діапазоні, який залежить від двох випадкових чисел a та b. Далі завдання ділилося на три етапи:
**а)** розрахувати статистику (середнє, мінімум, максимум, відхилення) для всієї матриці, рядків і стовпців;
**б)** привести всі значення до діапазону від 0 до 1;
**в)** розрахувати визначник і слід матриці $B^T B$.

**Для реалізації ми використали** метод `np.random.uniform` для генерації матриці дійсних чисел. У виводі ми спочатку друкуємо саму матрицю $A$, щоб бачити згенеровані дані. Статистику рахували через `np.mean`, `np.min`, `np.max` та `np.std`, керуючи напрямком обчислень через осі (`axis=1` для рядків, `axis=0` для стовпців). Нормалізацію (пункт б) зробили однією математичною операцією для всього масиву. Для пункту в) ми застосували транспонування `.T` і перемноження матриць `@`, а потім знайшли визначник і слід вбудованими функціями лінійної алгебри.

```python
a = np.random.randint(0, 11)
b = np.random.randint(0, 11)
low = min(a, b)
high = max(a, b) + 1

A = np.random.uniform(low, high, (20, 5))

mean_all = np.mean(A)
min_all = np.min(A)
max_all = np.max(A)
std_all = np.std(A)

mean_rows = np.mean(A, axis=1)
mean_cols = np.mean(A, axis=0)

B = (A - np.min(A)) / (np.max(A) - np.min(A))

BtB = B.T @ B
det_BtB = np.linalg.det(BtB)
trace_BtB = np.trace(BtB)

print("ЗАВДАННЯ 2")
print(f"Згенеровані межі діапазону: від {low} до {high} (на основі a={a}, b={b})")
print("\nПОЧАТКОВА МАТРИЦЯ А (20x5):")
print(A)

print("\nа) РОЗРАХУНОК СТАТИСТИКИ:")
print(f"По всій матриці -> Середнє: {mean_all:.2f} | Мін: {min_all:.2f} | Макс: {max_all:.2f} | Відхилення: {std_all:.2f}")
print("\nСереднє значення для кожного з 20 рядків:")
print(mean_rows)
print("\nСереднє значення для кожного з 5 стовпців:")
print(mean_cols)

print("\nб) НОРМАЛІЗАЦІЯ ЕЛЕМЕНТІВ:")
print(f"Мінімальне значення у новій матриці B: {np.min(B)}")
print(f"Максимальне значення у новій матриці B: {np.max(B)}")
print("Перші 3 рядки нормалізованої матриці B для перевірки:")
print(B[:3])

print("\nв) ВИЗНАЧНИК І СЛІД:")
print(f"Визначник матриці B^T * B дорівнює: {det_BtB}")
print(f"Слід матриці B^T * B дорівнює: {trace_BtB}")
```

### Завдання 3. Системи лінійних рівнянь
**По завданню треба було** створити матрицю $A$ (4 на 4) і вектор $b$ (4 на 1) із випадковими числами. Далі необхідно:
**а)** вивести їх як систему рівнянь $Ax=b$ формулами LaTeX[;
**б)** розв'язати її правилом Крамера;
**в)** перевірити правильність розв'язку матричним множенням, через обернену матрицю та функцією `linalg.solve()`.

**Для реалізації ми використали** модуль `IPython.display` для рендерингу рівнянь у форматі LaTeX просто в комірці виводу. Розв'язок за Крамером реалізували за допомогою циклу: ми копіювали головну матрицю, замінювали потрібний стовпець на вектор-результат і ділили отриманий визначник на головний визначник системи. Перевірки виконані стандартними засобами підмодуля `np.linalg`.

```python
from IPython.display import display, Markdown, Math

A_mat = np.random.randint(0, 11, (4, 4))
b_vec = np.random.randint(0, 11, (4, 1))

display(Markdown("**ЗАВДАННЯ 3**"))
print("Матриця коефіцієнтів A (4x4):")
print(A_mat)
print("\nВектор вільних членів b (4x1):")
print(b_vec)
print("\nа) СИСТЕМА РІВНЯНЬ (у форматі LaTeX) ===")

latex_str = "\\begin{cases} "
for i in range(4):
    eq = " + ".join([f"{A_mat[i,j]}x_{j+1}" for j in range(4)])
    latex_str += eq + f" = {b_vec[i,0]} \\\\ "
latex_str += "\\end{cases}"
display(Math(latex_str))

det_A = np.linalg.det(A_mat)
x_cramer = np.zeros((4, 1))

if det_A != 0:
    for i in range(4):
        A_temp = A_mat.copy()
        A_temp[:, i] = b_vec[:, 0]
        x_cramer[i, 0] = np.linalg.det(A_temp) / det_A

check_mult = A_mat @ x_cramer
check_inv = np.linalg.inv(A_mat) @ b_vec if det_A != 0 else np.zeros((4, 1))
check_solve = np.linalg.solve(A_mat, b_vec) if det_A != 0 else np.zeros((4, 1))

print("б) РОЗВ'ЯЗОК ЗА ПРАВИЛОМ КРАМЕРА (знайдений вектор x) ===")
print(x_cramer)

print("\nв) ПЕРЕВІРКА ПРАВИЛЬНОСТІ ===")
print("(i) Перевірка матричним множенням A * x (результат має точно збігатися з вектором b):")
print(check_mult)
print("\n(ii) Перевірка через обернену матрицю (x = A^-1 * b):")
print(check_inv)
print("\n(iii) Перевірка вбудованою функцією linalg.solve():")
print(check_solve)
```

### Завдання 4. Добуток Кронекера
**По завданню треба було** написати програму для обчислення добутку Кронекера двох матриць $A$ та $B$. Суть операції полягає в тому, що кожен окремий елемент першої матриці множиться на всю другу матрицю, утворюючи велику блочну структуру.

**Для реалізації ми використали** дві невеликі тестові матриці 2х2 для наочності. Сама операція виконується однією стандартною функцією `np.kron` із бібліотеки NumPy, яка бере на себе всю логіку розширення розмірностей і блочного множення. У вивід ми передали обидві початкові матриці та фінальний результат для порівняння.

```python
A_kr = np.random.randint(1, 5, (2, 2))
B_kr = np.random.randint(1, 5, (2, 2))

kronecker_product = np.kron(A_kr, B_kr)

print("ЗАВДАННЯ 4")
print("Вхідна матриця A (2x2):")
print(A_kr)
print("\nВхідна матриця B (2x2):")
print(B_kr)
print("\nРЕЗУЛЬТАТ")
print("Добуток Кронекера матриць A та B (розширена матриця 4x4):")
print(kronecker_product)
```

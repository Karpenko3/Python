### 1. Одновимірні масиви (вектори)
**Постановка задачі (Варіант 3):**
Нехай задано список різних випадкових чисел $A = [a_{1}, a_{2}, \dots, a_{n}]$. Масив формується за правилом (б): значення генеруються як цілі числа в діапазоні від -10 до 10 включно. Необхідно розробити програму для формування іншого списку, в якому елементи переставлені таким чином, щоб спочатку були розташовані всі невід'ємні елементи ($a_{i} \ge 0$), а вкінці — від'ємні ($a_{i} < 0$).

**Детальний опис алгоритму та роботи структур даних:**
1. **Ініціалізація та генерація:** Для створення початкового масиву використовується генератор списків (List Comprehension). Цей підхід у Python працює швидше та є більш ресурсоефективним за звичайний цикл `for`.
2. **Розділення елементів (Фільтрація):** Нам потрібно згрупувати елементи виключно за знаком. Ми уникаємо високовартісних алгоритмів сортування ($O(N \log N)$), натомість застосовуємо лінійну фільтрацію ($O(N)$). Створюються два окремі підмасиви за умовами `x >= 0` та `x < 0`.
3. **Конкатенація (Злиття):** За допомогою оператора `+` ми об'єднуємо обидва підмасиви у фінальний список.

```python
import random

n = 15
print(f"Кількість елементів у списку: n = {n}\n")


original_array = [random.randint(-10, 10) for _ in range(n)]
print("Проміжний результат: Початковий масив ---")
print(original_array)

non_negative_elements = [x for x in original_array if x >= 0]
negative_elements = [x for x in original_array if x < 0]

new_array = non_negative_elements + negative_elements

print("\nФінальний результат: Перегрупований масив ---")
print(new_array)
```

### 2. Двовимірні масиви (матриці)
**Постановка задачі (Варіант 3):**
Необхідно згенерувати двовимірний список $N \times M$ (матрицю), де кількість рядків $N$ визначається кількістю літер в імені ($N=5$ для "Артем"), а кількість стовпців $M$ — кількістю літер у прізвищі ($M=8$ для "Карпенко"). Елементи матриці формуються за правилом (б) — це випадкові цілі числа з діапазону $[-10, 10]$.
Завдання варіанта: знайти рядок, який містить мінімальний додатний елемент матриці, та змінити його місцями з останнім рядком.

**Детальний опис алгоритму та роботи з пам'яттю:**
1. **Створення двовимірної структури:** Двовимірний масив ініціалізується за допомогою вкладених генераторів: зовнішній відповідає за створення рядків, внутрішній — за наповнення стовпців.
2. **Лінійний пошук екстремуму:** Пошук найменшого числа серед виключно додатних елементів відбувається через подвійний цикл з перевіркою умови $0 < x < \text{min\_positive}$.
3. **Механіка швидкого обміну (Swap):** Конструкція `matrix[target_row_idx], matrix[-1] = matrix[-1], matrix[target_row_idx]` виконується миттєво (за час $O(1)$). Python не переміщує самі числа у пам'яті, а лише перепризначає посилання (вказівники) на вкладені списки.

```python

n_rows = 5  # Кількість літер в імені "Артем"
m_cols = 8  # Кількість літер в прізвищі "Карпенко"

# 2. Генерація матриці (випадкові числа від -10 до 10)
matrix = [[random.randint(-10, 10) for _ in range(m_cols)] for _ in range(n_rows)]

print("Проміжний результат: Початкова матриця ")
for row in matrix:
    print([f"{num:3}" for num in row])

# 3. Пошук мінімального додатного елемента та індексу його рядка
min_positive = float('inf')
target_row_idx = -1

for i in range(n_rows):
    for j in range(m_cols):
        if 0 < matrix[i][j] < min_positive:
            min_positive = matrix[i][j]
            target_row_idx = i

# 4. Перестановка рядків
if target_row_idx != -1:
    print(f"\nЗнайдено мінімальний додатний елемент: {min_positive}")
    print(f"Він знаходиться у рядку з індексом: {target_row_idx}")
    
    # Змінюємо вказівники на рядки місцями
    matrix[target_row_idx], matrix[-1] = matrix[-1], matrix[target_row_idx]
    
    print("\n Фінальний результат: Матриця після перестановки рядків")
    for row in matrix:
        print([f"{num:3}" for num in row])
else:
    # Обробка виключення, якщо згенеровано матрицю лише з від'ємними числами та нулями
    print("\nУ згенерованій матриці відсутні додатні елементи. Перестановка не виконувалась.")
```

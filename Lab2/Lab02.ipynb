### 1. Моделювання гри "Герб-решка" (Початкові умови)
Згідно з Варіантом 4, початковий капітал першого гравця становить $m = 4$, а другого — $n = 5$. 
Два гравці грають у "герб-решка". Кожен раунд вони підкидають монету і за результатами кидання один виграє 1 грошову одиницю у іншого. Гра триває, поки хтось не розориться (капітал = 0).

Щоб програмно реалізувати цей процес і промоделювати результат 1000 таких ігор, ми використаємо:
*   Функцію користувача `simulate_games`, яка прийматиме початкові капітали.
*   Цикл `for` для повторення моделювання 1000 разів.
*   Цикл `while` для відтворення самої гри раунд за раундом, поки баланс обох гравців залишається додатнім.
*   Функцію `random.choice([0, 1])` з бібліотеки `random` для імітації випадкового кидання монети.

```python
import random

def simulate_games(m, n, num_games=1000):
    """
    Моделює серію ігор між двома гравцями до розорення одного з них.
    """
    p1_wins = 0
    p2_wins = 0
    total_rounds = 0
    
    for _ in range(num_games):
        p1_cap = m
        p2_cap = n
        rounds = 0
        
        while p1_cap > 0 and p2_cap > 0:
            toss = random.choice([0, 1])
            
            if toss == 0:
                p1_cap += 1
                p2_cap -= 1
            else:
                p1_cap -= 1
                p2_cap += 1
                
            rounds += 1
            
        total_rounds += rounds
        
        if p1_cap > 0:
            p1_wins += 1
        else:
            p2_wins += 1
            
    avg_rounds = total_rounds / num_games
    return p1_wins, p2_wins, avg_rounds

p1_wins_1, p2_wins_1, avg_rounds_1 = simulate_games(4, 5, 1000)

print("Результати моделювання (m = 4, n = 5):")
print(f"Перший гравець виграв: {p1_wins_1} разів")
print(f"Другий гравець виграв: {p2_wins_1} разів")
print(f"В середньому одна гра тривала: {avg_rounds_1:.1f} раундів")
```

Наступним кроком необхідно збільшити капітал першого гравця на одиницю і наново виконати завдання. Таким чином, тепер початковий капітал обох гравців стає рівним ($m = 5, n = 5$). 

Завдяки тому, що алгоритм гри ми винесли в окрему функцію `simulate_games`, нам не потрібно дублювати код. Достатньо викликати цю ж функцію з новими аргументами та проаналізувати зміну математичного очікування.

```python
# Виконання п.2: Збільшуємо капітал першого гравця на одиницю (m = 5, n = 5)
p1_wins_2, p2_wins_2, avg_rounds_2 = simulate_games(5, 5, 1000)

print("--- Результати моделювання (m = 5, n = 5):")
print(f"Перший гравець виграв: {p1_wins_2} разів")
print(f"Другий гравець виграв: {p2_wins_2} разів")
print(f"В середньому одна гра тривала: {avg_rounds_2:.1f} раундів")
```

**Аналіз впливу початкового капіталу:**

2. **Зрівняння шансів:** Збільшення капіталу першого гравця до $m=5$ призвело до вирівнювання шансів. При капіталах 5 проти 5 обидва гравці мають приблизно однакову ймовірність на перемогу (близько 500 перемог із 1000).
3. **Тривалість гри:** Збільшення загальної кількості грошей у грі закономірно впливає на її тривалість.
